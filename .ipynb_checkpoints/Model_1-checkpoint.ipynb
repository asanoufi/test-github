```python
#--------------------
###MODELLING SETTINGS
#--------------------


##Outcome being predicted:
#102 = 2% slower than comparison best
#99 = 1% faster than comparison best
OUTCOME_VARIABLE = "pct_from_season_best"

#Number of cross validation folds
CV_FOLDS = 5

#Ensures reproducible MLP results
RANDOM_STATE = 42

#Race characteristics that may explain differences in relative performance.
CONTEXT_VARIABLES = [
    "RaceEventSwimmingStyleName",
    "RaceEventDistance",
    "RacePhaseGroup",]

#Columns used to determine whether the selected monitoring datasets genuinely covered a race.

#They define the cohort before occasional missing measurements are imputed.

#E.g. distance_40d confirms complete Commit history, cmj_tests_40d confirms ForceDecks coverage, 
#even when zero tests occurred in that window
REQUIRED_COVERAGE_VARIABLES = [
    "distance_40d",
    "cmj_tests_40d",]

#Train Split, proportion of chronological race dates used before the first validation period begins.
INITIAL_TRAIN_FRACTION = 0.40


#Compare three feature specifications on the same race cohort.

#context_only: how well race characteristics predict the outcome.
#monitoring_only: tests the selected monitoring features.
#combined: tests whether monitoring adds value alongside race context.
MODEL_FEATURE_SETS = {
    "context_only": CONTEXT_VARIABLES,
    "monitoring_only": INPUT_VARIABLES,
    "combined": CONTEXT_VARIABLES + INPUT_VARIABLES,}

print("Number of selected monitoring variables:", len(INPUT_VARIABLES))

for model_name, variables in MODEL_FEATURE_SETS.items():
    print(
        f"{model_name}: {len(variables)} variables")
```

```python
#Load in and clean races

cols_needed = [
    "RaceId",
    "Date",
    "AthleteName",
    "AthleteGenderName",
    "RaceEventName",
    "RaceEventDistance",
    "RaceEventSwimmingStyleName",
    "RaceEventIsRelay",
    "PoolLength",
    "PhaseName",
    "RaceTime",
    "CompetitionName",
]

df_nemo = pd.read_csv(
    NEMO_FILE,
    usecols=cols_needed,
    parse_dates=["Date"],
    dayfirst = True,
    low_memory=False
)

#Create unique identifier for every row
df_nemo["ModelRaceId"] = np.arange(len(df_nemo))

print(f"Original dataset: {df_nemo.shape}")
print(f"Missing RaceId: {df_nemo['RaceId'].isna().sum()}")
print(f"Missing CompetitionName: {df_nemo['CompetitionName'].isna().sum()}")
```

```python
df_lc = df_nemo.copy()

#Keep individual races with valid race times
df_lc = df_lc[df_lc["RaceEventIsRelay"] == False]
df_lc = df_lc[df_lc["RaceTime"] > 0]

#Remove unwanted competitions and known error races
df_lc = df_lc[~df_lc["CompetitionName"].isin(JUNK_COMPETITIONS)]
df_lc = df_lc[~df_lc["RaceId"].isin(BAD_RACE_IDS)]

#Keep long-course races
df_lc = df_lc[df_lc["PoolLength"] == 50].copy()

print(f"Clean long-course dataset: {df_lc.shape}")
print(f"Rows removed: {len(df_nemo) - len(df_lc)}")
```

```python
def assign_season(date):
    if pd.isna(date):
        return pd.NA

    if date.month >= 9:
        start_year = date.year
    else:
        start_year = date.year - 1

    return f"{start_year}_{str(start_year + 1)[-2:]}"


df_lc["SeasonId"] = df_lc["Date"].apply(assign_season)

#Put races into the correct chronological order
df_lc = df_lc.sort_values(
    ["AthleteName", "RaceEventName", "Date", "ModelRaceId"]
).reset_index(drop=True)

print(df_lc["SeasonId"].value_counts(dropna=False).sort_index())
print(f"\nMissing dates: {df_lc['Date'].isna().sum()}")
```

```python
def expanding_min_excluding_current(series):
    return series.shift(1).expanding().min()


#Best time achieved before each race in the current season
df_lc["season_best_so_far"] = (
    df_lc
    .groupby(
        ["AthleteName", "RaceEventName", "SeasonId"],
        sort=False
    )["RaceTime"]
    .transform(expanding_min_excluding_current))


#Full season best for each athlete and event
season_bests = (
    df_lc
    .groupby(
        ["AthleteName", "RaceEventName", "SeasonId"],
        as_index=False
    )["RaceTime"]
    .min()
    .rename(columns={"RaceTime": "full_season_best"}))


def previous_season(season_id):
    start_year = int(season_id.split("_")[0])
    previous_start = start_year - 1
    return f"{previous_start}_{str(start_year)[-2:]}"


season_bests["PreviousSeasonId"] = (
    season_bests["SeasonId"].apply(previous_season))


#Match each season to the immediately preceding season
prior_lookup = season_bests.merge(
    season_bests[
        [
            "AthleteName",
            "RaceEventName",
            "SeasonId",
            "full_season_best",
        ]
    ],
    left_on=[
        "AthleteName",
        "RaceEventName",
        "PreviousSeasonId",
    ],
    right_on=[
        "AthleteName",
        "RaceEventName",
        "SeasonId",
    ],
    how="left",
    suffixes=("", "_prior"),
)

prior_lookup = prior_lookup[
    [
        "AthleteName",
        "RaceEventName",
        "SeasonId",
        "full_season_best_prior",
    ]
].rename(
    columns={"full_season_best_prior": "prior_season_best"}
)


#Add the preceding season's best to each race
df_lc = df_lc.merge(
    prior_lookup,
    on=["AthleteName", "RaceEventName", "SeasonId"],
    how="left",
)

#Restore chronological order after merging
df_lc = df_lc.sort_values(
    ["AthleteName", "RaceEventName", "Date", "ModelRaceId"]
).reset_index(drop=True)


#Prefer the current season's best; otherwise use the prior season
df_lc["comparison_time"] = (
    df_lc["season_best_so_far"]
    .fillna(df_lc["prior_season_best"])
)

df_lc["comparison_source"] = np.select(
    [
        df_lc["season_best_so_far"].notna(),
        df_lc["prior_season_best"].notna(),
    ],
    [
        "current_season",
        "prior_season",
    ],
    default="none",
)


print(df_lc["comparison_source"].value_counts())
print(
    "\nRaces without a valid comparison:",
    df_lc["comparison_time"].isna().sum()
)
```

```python
#Create the outcome variable

df_lc["pct_from_season_best"] = (
    df_lc["RaceTime"] / df_lc["comparison_time"]
) * 100

df_lc["is_new_season_best"] = (
    df_lc["pct_from_season_best"] < 100
)

# Keep only races with a calculable outcome
df_model = df_lc[df_lc["comparison_time"].notna()].copy()

print(f"Final outcome dataset: {df_model.shape}")
print(df_model["pct_from_season_best"].describe())

print(
    "\nValues outside 90–110:",
    (
        (df_model["pct_from_season_best"] < 90)
        | (df_model["pct_from_season_best"] > 110)
    ).sum()
)
```

```python
##Commit Features Function

def create_training_features(
    commit,
    athlete_name,
    race_date,
    race_stroke
):
    """
    Summarise an athlete's pool training before one race.

    Creates features over each period specified in
    TRAINING_WINDOWS (10d, 40d).
    """

    #Convert the race date to a UTC timestamp so it can be compared
    #with the UTC training timestamps without a timezone error
    
    race_timestamp = pd.Timestamp(race_date)

    if race_timestamp.tzinfo is None:
        race_timestamp = race_timestamp.tz_localize("UTC")
    else:
        race_timestamp = race_timestamp.tz_convert("UTC")

    #Select only the Commit rows belonging to this athlete.
    athlete_training = commit[
        commit["About"] == athlete_name
    ]

    features = {}

    #If the athlete has no Commit data, return missing values rather
    #than zeros. Zero would incorrectly imply that they did no training.
    athlete_has_data = not athlete_training.empty

    for days in TRAINING_WINDOWS:

        #Create the boundaries of the pre-race window.
        #The race date itself is excluded.
        window_start = race_timestamp - pd.Timedelta(days=days)

        #A window is considered usable only if the athlete has recorded
        #training data extending back to the beginning of that window.
        complete_window = (
            athlete_has_data
            and athlete_training["TrainingTimestamp"].min() <= window_start
        )

        if complete_window:

            #Select this athlete's sessions occurring within the window.
            window_data = athlete_training[
                (athlete_training["TrainingTimestamp"] >= window_start)
                & (athlete_training["TrainingTimestamp"] < race_timestamp)
            ]

            #Sum each selected training-load measure over the window.
            for column in TRAINING_LOAD_VARIABLES:

                #Give total distance "distance_10d" rather than "workout_total_distance_10d".
                feature_name = (
                    "distance"
                    if column == "workout_total_distance"
                    else column
                )

                features[f"{feature_name}_{days}d"] = (
                    window_data[column].sum()
                )

            if INCLUDE_SESSION_COUNT:
                #nunique() counts two same-day sessions separately when they have different workout IDs.
                features[f"sessions_{days}d"] = (
                    window_data["workout_id"].nunique())

            if INCLUDE_STROKE_DISTANCE:
                #Select the Commit stroke column corresponding to the stroke of the race being predicted.
                stroke_column = STROKE_MAP.get(race_stroke)

                if stroke_column in window_data.columns:
                    features[f"race_stroke_distance_{days}d"] = (
                        window_data[stroke_column].sum())
                else:
                    features[f"race_stroke_distance_{days}d"] = np.nan

        else:
            #An incomplete or unavailable window is missing—not zero.
            for column in TRAINING_LOAD_VARIABLES:
                feature_name = (
                    "distance"
                    if column == "workout_total_distance"
                    else column
                )
                features[f"{feature_name}_{days}d"] = np.nan

            if INCLUDE_SESSION_COUNT:
                features[f"sessions_{days}d"] = np.nan

            if INCLUDE_STROKE_DISTANCE:
                features[f"race_stroke_distance_{days}d"] = np.nan

    #Calculate taper ratio only when both selected windows are available.
    if INCLUDE_TAPER_RATIO:
        distance_10d = features.get("distance_10d")
        distance_40d = features.get("distance_40d")

        if (
            pd.notna(distance_10d)
            and pd.notna(distance_40d)
            and distance_40d > 0
        ):
            features["taper_ratio"] = (
                distance_10d / distance_40d
            ) * 100
        else:
            features["taper_ratio"] = np.nan

    return features
```

```python
##Run the training load feature function once for every race in df_model.

#axis=1 means that the function is applied row by row rather than separately to each dataframe column.
training_features = df_model.apply(
    lambda row: pd.Series(
        create_training_features(
            commit=df_commit,
            athlete_name=row["AthleteName"],
            race_date=row["Date"],
            race_stroke=row["RaceEventSwimmingStyleName"],
        )
    ),
    axis=1,
)

##Add the newly calculated training features to the race dataset.

#pd.concat(..., axis=1) joins the two dataframes side by side while preserving the corresponding race-row index.
df_model = pd.concat(
    [df_model, training_features],
    axis=1
)

print(f"Dataset after adding Commit features: {df_model.shape}")

##Display the number of usable and missing observations for each newly created training feature.
training_feature_coverage = pd.DataFrame({
    "available": training_features.notna().sum(),
    "missing": training_features.isna().sum(),
})

print(training_feature_coverage)
```

```python
def aggregate_forcedecks_trials(data, test_type, variables):
    """
    Create one row per ForceDecks testing occasion while retaining
    both the mean and maximum result across its individual trials.
    """

    #Retain the requested test type and relevant columns.
    test_data = data[
        data["Test Type"] == test_type
    ][
        [
            "Id",
            "About",
            "ForceDecksTimestamp",
            *variables,
        ]
    ].copy()

    #Convert the selected measurements to numeric values.
    #Invalid text entries become NaN and are ignored in calculations.
    for column in variables:
        test_data[column] = pd.to_numeric(
            test_data[column],
            errors="coerce"
        )

    grouping_columns = [
        "Id",
        "About",
        "ForceDecksTimestamp",
    ]

    #For each test ID, calculate both the average trial result
    #and the highest trial result for every selected variable.
    test_measurements = (
        test_data
        .groupby(grouping_columns)[variables]
        .agg(["mean", "max"])
    )

    #Convert column names into simple names such as:
    test_measurements.columns = [
        f"{variable}__{summary}"
        for variable, summary in test_measurements.columns
    ]

    #Return the grouping columns to normal dataframe columns.
    test_level_data = test_measurements.reset_index()

    return test_level_data


#Produce one row per CMJ testing occasion.
fd_cmj_tests = aggregate_forcedecks_trials(
    data=df_forcedecks,
    test_type="CMJ",
    variables=CMJ_VARIABLES,
)

#Produce one row per shoulder-isometric testing occasion.
fd_shoulder_tests = aggregate_forcedecks_trials(
    data=df_forcedecks,
    test_type="SHLDISOI",
    variables=SHOULDER_ISO_VARIABLES,
)

print(f"CMJ testing occasions: {len(fd_cmj_tests)}")
print(f"Shoulder testing occasions: {len(fd_shoulder_tests)}")
```

```python
import re  #(regular expression module, cleans the variable names)

def simplify_variable_name(variable_name):
    """
    Convert a long ForceDecks heading into a readable Python column name.

    Example:
    "Peak Power / BM [W/kg]" becomes "peak_power_bm".
    """

    #Remove units enclosed in square brackets.
    simplified = re.sub(r"\[[^\]]*\]", "", variable_name)

    #Keep useful text inside normal brackets but remove the brackets.
    simplified = simplified.replace("(", "").replace(")", "")

    #Replace symbols and spaces with underscores.
    simplified = simplified.replace("/", " ")
    simplified = re.sub(r"[^A-Za-z0-9]+", "_", simplified)

    #Remove underscores from the beginning/end and use lowercase.
    return simplified.strip("_").lower()


def create_forcedecks_features(
    test_data,
    athlete_name,
    race_date,
    variables,
    test_prefix,
):
    """
    Summarise ForceDecks results recorded before one race.
    """

    #Convert the race date into a UTC timestamp so it can be compared with ForceDecks Timestamp.
    race_timestamp = pd.Timestamp(race_date)

    if race_timestamp.tzinfo is None:
        race_timestamp = race_timestamp.tz_localize("UTC")
    else:
        race_timestamp = race_timestamp.tz_convert("UTC")

    #Keep testing occasions belonging to the race athlete.
    athlete_tests = test_data[
        test_data["About"] == athlete_name
    ]

    features = {}

    #Identify when this particular ForceDecks test dataset begins.
    #This is different for CMJ and shoulder-iso testing.
    data_start = test_data["ForceDecksTimestamp"].min()

    for days in FORCEDECKS_WINDOWS:

        #Define beginning of the pre-race window.
        window_start = race_timestamp - pd.Timedelta(days=days)

        #A window is unavailable if the athlete has no data for this test
        #or if ForceDecks collection had not started by the window beginning.
        window_is_available = (
            not athlete_tests.empty
            and pd.notna(data_start)
            and data_start <= window_start
        )

        if not window_is_available:
            features[f"{test_prefix}_tests_{days}d"] = np.nan

            for variable in variables:
                short_name = simplify_variable_name(variable)

                features[
                    f"{test_prefix}_{short_name}_mean_{days}d"
                ] = np.nan

                features[
                    f"{test_prefix}_{short_name}_max_{days}d"
                ] = np.nan

            #Move directly to the next window without calculating false zeros.
            continue

        #Keep testing occasions inside the window while excluding
        #tests performed on the race date itself.
        window_tests = athlete_tests[
            (athlete_tests["ForceDecksTimestamp"] >= window_start)
            & (athlete_tests["ForceDecksTimestamp"] < race_timestamp)
        ]

        #Count actual testing occasions, not individual trials.
        features[f"{test_prefix}_tests_{days}d"] = len(window_tests)

        for variable in variables:

            short_name = simplify_variable_name(variable)

            #These columns were created when the trials within each
            #ForceDecks test ID were aggregated.
            test_mean_column = f"{variable}__mean"
            test_max_column = f"{variable}__max"

            #Calculate the average of the testing-occasion means.
            #If there are no tests, pandas returns NaN.
            features[
                f"{test_prefix}_{short_name}_mean_{days}d"
            ] = window_tests[test_mean_column].mean()

            #Calculate the highest trial maximum recorded across all
            #testing occasions in the pre-race window.
            features[
                f"{test_prefix}_{short_name}_max_{days}d"
            ] = window_tests[test_max_column].max()

    return features
```

```python
##Add CMJ and Shoulder ISO to each race

def create_cmj_features_for_race(race_row):
    """
    Calculate CMJ features for one race.

    race_row contains all information from one row of df_model,
    including the athlete identifier and race date.
    """

    return pd.Series(
        create_forcedecks_features(
            #Use the dataset containing one row per CMJ test.
            test_data=fd_cmj_tests,

            #Match ForceDecks "About" to the race athlete.
            athlete_name=race_row["AthleteName"],

            #Use the race date as the end of each modelling window.
            race_date=race_row["Date"],

            #Calculate features only for the CMJ variables selected
            #in the settings section at the top of the notebook.
            variables=CMJ_VARIABLES,

            #Add "cmj" to the beginning of every generated feature.
            test_prefix="cmj",
        )
    )


def create_shoulder_features_for_race(race_row):
    """
    Calculate shoulder-isometric features for one race.
    """

    return pd.Series(
        create_forcedecks_features(
            #Use the dataset containing one row per shoulder test.
            test_data=fd_shoulder_tests,

            #Match the race athlete to the ForceDecks athlete.
            athlete_name=race_row["AthleteName"],

            #Include only testing completed before this race date.
            race_date=race_row["Date"],

            #Use the shoulder variables selected in the settings.
            variables=SHOULDER_ISO_VARIABLES,

            #Add "shoulder" to each generated feature name.
            test_prefix="shoulder",
        )
    )


#Apply the CMJ function separately to every race.

cmj_features = df_model.apply(
    create_cmj_features_for_race,
    axis=1,                   #tells pandas to work across rows instead of separately to columns
)


#Repeat the row-by-row calculation for shoulder-isometric testing.
shoulder_features = df_model.apply(
    create_shoulder_features_for_race,
    axis=1,
)


#Add the new feature columns beside the existing race and Commit columns.

df_model = pd.concat(
    [
        df_model,
        cmj_features,
        shoulder_features,
    ],
    axis=1,           #Joins side by side using the existing df indexes
)


print(f"Dataset after adding ForceDecks features: {df_model.shape}")

#Count races for which at least one actual testing occasion occurred
#within each pre-race window. NaN and zero are not counted.
print(
    "Races with at least one CMJ in previous 10 days:",
    (df_model["cmj_tests_10d"] > 0).sum()
)

print(
    "Races with at least one CMJ in previous 40 days:",
    (df_model["cmj_tests_40d"] > 0).sum()
)

print(
    "Races with at least one shoulder test in previous 10 days:",
    (df_model["shoulder_tests_10d"] > 0).sum()
)

print(
    "Races with at least one shoulder test in previous 40 days:",
    (df_model["shoulder_tests_40d"] > 0).sum()
)
```

```python
##Define HRV rolling feature function

def create_hrv_features(
    hrv_data,
    athlete_name,
    race_date,
):
    """
    Calculate rolling HRV and sleep features before one race.
    """

    #Convert the race date to midnight without a timezone because HRVDate
    #also represents a date without a specific time.
    race_day = pd.Timestamp(race_date).normalize()

    #Keep only daily records belonging to this athlete.
    athlete_hrv = hrv_data[
        hrv_data["About"] == athlete_name
    ]

    #Identify the overall period covered by the HRV dataset.
    data_start = hrv_data["HRVDate"].min()
    data_end = hrv_data["HRVDate"].max()

    features = {}

    for days in HRV_WINDOWS:

        #If race-day HRV is included, a 3-day window consists of the
        #race day and the two preceding calendar days.
        window_end = (
            race_day
            if INCLUDE_RACE_DAY_HRV
            else race_day - pd.Timedelta(days=1)
        )

        window_start = (
            window_end - pd.Timedelta(days=days - 1)
        )

        #Check that the athlete exists in the HRV dataset and that the
        #dataset covers the entire requested calendar window.
        complete_window = (
            not athlete_hrv.empty
            and data_start <= window_start
            and data_end >= window_end
        )

        for variable in HRV_VARIABLES:

            #Convert names such as "Sleep Value" into "sleep_value".
            short_name = simplify_variable_name(variable)

            observation_name = (
                f"{short_name}_observations_{days}d"
            )
            mean_name = f"{short_name}_mean_{days}d"

            if not complete_window:
                #Missing coverage is recorded as NaN rather than zero.
                features[observation_name] = np.nan
                features[mean_name] = np.nan
                continue

            #Select this athlete's daily records inside the window.
            window_data = athlete_hrv[
                (athlete_hrv["HRVDate"] >= window_start)
                & (athlete_hrv["HRVDate"] <= window_end)
            ]

            #Count only non-missing observations for this variable.
            valid_observations = window_data[variable].dropna()

            observation_count = len(valid_observations)
            features[observation_name] = observation_count

            #Calculate the mean only when the window contains the
            #minimum number of valid observations specified above.
            if observation_count >= HRV_MIN_OBSERVATIONS[days]:
                features[mean_name] = valid_observations.mean()
            else:
                features[mean_name] = np.nan

    #Express recent HRV relative to the athlete's 28-day value.
    #Zero means the 7- and 28-day means are identical.
    hrv_7d = features.get("hrv_mean_7d")
    hrv_28d = features.get("hrv_mean_28d")

    if (
        pd.notna(hrv_7d)
        and pd.notna(hrv_28d)
        and hrv_28d != 0
    ):
        features["hrv_7d_vs_28d_pct"] = (
            (hrv_7d / hrv_28d) - 1
        ) * 100
    else:
        features["hrv_7d_vs_28d_pct"] = np.nan

    return features
```

```python
##Apply HRV function to races

def create_hrv_features_for_race(race_row):
    """
    Extract athlete identifier and race date from one race row,
    calculate the corresponding pre-race HRV features.
    """

    return pd.Series(
        create_hrv_features(
            #Use the prepared daily HRV dataset.
            hrv_data=df_hrv,

            #Match HRV "About" to Nemo "AthleteName".
            athlete_name=race_row["AthleteName"],

            #Use this race's date to define the rolling windows.
            race_date=race_row["Date"],
        )
    )


#Apply the function separately to every row in the race dataset.

hrv_features = df_model.apply(
    create_hrv_features_for_race,
    axis=1,        #tells pandas that each item passed to function should be one complete race
)


#Join the generated HRV columns beside the existing race, Commit and ForceDecks columns.
df_model = pd.concat(
    [
        df_model,
        hrv_features,
    ],
    axis=1,
)


print(f"Dataset after adding HRV features: {df_model.shape}")

#Report how many races have usable values for the main HRV features.
print(
    "Races with a valid 3-day HRV mean:",
    df_model["hrv_mean_3d"].notna().sum()
)

print(
    "Races with a valid 7-day HRV mean:",
    df_model["hrv_mean_7d"].notna().sum()
)

print(
    "Races with a valid 28-day HRV mean:",
    df_model["hrv_mean_28d"].notna().sum()
)

print(
    "Races with a valid 7-day versus 28-day HRV comparison:",
    df_model["hrv_7d_vs_28d_pct"].notna().sum()
)

print(
    "Races with a valid 7-day sleep mean:",
    df_model["sleep_value_mean_7d"].notna().sum()
)
```

```python
##Define race-matching function

def create_iap_features(
    iap_data,
    athlete_name,
    race_date,
):
    """
    Match one race to the athlete's most recent valid IAP assessment. (given it is <14 days old)
    """

    #Convert the race date into a normalised calendar date.
    race_day = pd.Timestamp(race_date).normalize()

    #Keep only IAP records belonging to this athlete.
    athlete_iap = iap_data[
        iap_data["About"] == athlete_name
    ]

    #Keep assessments completed strictly before the race date.
    
    previous_assessments = athlete_iap[
        athlete_iap["IAPDate"] < race_day
    ]

    feature_names = [
        "iap_training_concern",
        "iap_health_concern",
        "iap_general_concern",
        "iap_total_concern",
        "iap_worst_status",
        "iap_days_since_assessment",
    ]

    #If the athlete has no pre-race assessment, all IAP features are unavailable for this race.
    if previous_assessments.empty:
        return {
            feature_name: np.nan
            for feature_name in feature_names
        }

    #Sort the previous records by date and select the most recent one.
    latest_assessment = (
        previous_assessments
        .sort_values("IAPDate")
        .iloc[-1]
    )

    #Calculate the number of calendar days between the assessment and the race.
    days_since_assessment = (
        race_day - latest_assessment["IAPDate"]
    ).days

    #Do not use an assessment older than the maximum age selected in the settings section.
    if days_since_assessment > IAP_MAX_AGE_DAYS:
        return {
            feature_name: np.nan
            for feature_name in feature_names
        }

    return {
        "iap_training_concern":
            latest_assessment["training_concern_score"],

        "iap_health_concern":
            latest_assessment["health_concern_score"],

        "iap_general_concern":
            latest_assessment["general_concern_score"],

        "iap_total_concern":
            latest_assessment["iap_total_concern"],

        "iap_worst_status":
            latest_assessment["iap_worst_status"],

        "iap_days_since_assessment":
            days_since_assessment,
    }
```

```python
##Apply IAP function to races

def create_iap_features_for_race(race_row):
    """
    Extract the athlete and date from one race row, then match it
    to the athlete's most recent valid weekly IAP assessment.
    """

    return pd.Series(
        create_iap_features(
            #Use the prepared and numerically encoded IAP dataset.
            iap_data=df_iap,

            #Match IAP "About" to Nemo "AthleteName".
            athlete_name=race_row["AthleteName"],

            #Use this race date to identify the latest prior assessment.
            race_date=race_row["Date"],
        )
    )


#Apply the matching function separately to every race row.
iap_features = df_model.apply(
    create_iap_features_for_race,
    axis=1,
)


#Add the resulting IAP columns beside the existing model features.
df_model = pd.concat(
    [
        df_model,
        iap_features,
    ],
    axis=1,
)


print(f"Dataset after adding IAP features: {df_model.shape}")

#Count races with a valid recent IAP assessment.
print(
    "Races with IAP coverage:",
    df_model["iap_total_concern"].notna().sum()
)

#Show how recent the matched assessments were.
print(
    "\nDays between IAP assessment and race:"
)

print(
    df_model.loc[
        df_model["iap_days_since_assessment"].notna(),
        "iap_days_since_assessment"
    ].describe()
)

#Display the distribution of the overall weekly concern score among
#races with valid IAP coverage.
print(
    "\nIAP total concern distribution:"
)

print(
    df_model["iap_total_concern"]
    .value_counts(dropna=False)
    .sort_index()
)
```

```python
##Aggregate efforts in each round (best/average of efforts)

def aggregate_skill_efforts(
    data,
    variables,
    stroke_column,
):
    """
    Create one row per athlete-date-stroke testing occasion.

    Each occasion contains two or three individual efforts. Both the
    average time and fastest time are kept for modelling.
    """

    grouping_columns = [
        "Name",
        "SkillsDate",
        stroke_column,
    ]

    #Group efforts belonging to the same athlete, date and stroke.
    
    #mean = average performance across efforts
    #min  = fastest performance
    aggregated = (
        data
        .groupby(grouping_columns)[variables]
        .agg(["mean", "min"])
    )

    #Replace the two-level column headings created by aggregation with simple names such as:
    #start_10m_out_mean
    #start_10m_out_best
    new_column_names = []

    for variable, summary in aggregated.columns:
        short_name = simplify_variable_name(variable)

        #Rename "min" as "best" to make its meaning clearer.
        summary_name = (
            "best"
            if summary == "min"
            else "mean"
        )

        new_column_names.append(
            f"{short_name}_{summary_name}"
        )

    aggregated.columns = new_column_names

    #Return the athlete, date and stroke identifiers to normal columns.
    aggregated = aggregated.reset_index()

    #Give the stroke column the same name in the start and turn datasets.
    aggregated = aggregated.rename(
        columns={stroke_column: "SkillStroke"}
    )

    return aggregated


#Create one row per start-testing occasion.
skills_start_tests = aggregate_skill_efforts(
    data=df_skills,
    variables=START_SKILL_VARIABLES,
    stroke_column="StartStrokeClean",
)


#Create one row per turn-testing occasion.
skills_turn_tests = aggregate_skill_efforts(
    data=df_skills,
    variables=TURN_SKILL_VARIABLES,
    stroke_column="TurnStrokeClean",
)


print(f"Start testing occasions: {len(skills_start_tests)}")
print(f"Turn testing occasions: {len(skills_turn_tests)}")

print("\nStart feature columns:")
print(skills_start_tests.columns.tolist())

print("\nTurn feature columns:")
print(skills_turn_tests.columns.tolist())
```

```python
##Define pre-race latest-test function

def find_latest_skill_test(
    test_data,
    athlete_name,
    race_date,
    target_stroke,
    feature_prefix,
):
    """
    Find an athlete's most recent stroke-matched skill test before a race.
    """

    #Convert the race date into a normalised calendar date.
    race_day = pd.Timestamp(race_date).normalize()

    #Identify measurement columns by excluding the three columns used to identify athlete, test date and stroke.
    identifier_columns = {
        "Name",
        "SkillsDate",
        "SkillStroke",
    }

    measurement_columns = [
        column
        for column in test_data.columns
        if column not in identifier_columns
    ]

    #Include a feature recording how old the matched assessment is.
    days_since_feature = (
        f"{feature_prefix}_days_since_test"
    )

    output_columns = [
        *measurement_columns,
        days_since_feature,
    ]

    #Retain tests belonging to this athlete and matching the required race stroke.
    matching_tests = test_data[
        (test_data["Name"] == athlete_name)
        & (test_data["SkillStroke"] == target_stroke)
        & (test_data["SkillsDate"] < race_day)
    ]

    #The race date itself is excluded 
    if matching_tests.empty:
        return {
            column: np.nan
            for column in output_columns
        }

    #Sort chronologically and retain only the latest eligible test.
    latest_test = (
        matching_tests
        .sort_values("SkillsDate")
        .iloc[-1]
    )

    #Calculate the number of days between the skill test and race.
    days_since_test = (
        race_day - latest_test["SkillsDate"]
    ).days

    #Reject the test if it is older than the maximum age specified in the settings section.
    if days_since_test > SKILLS_MAX_AGE_DAYS:
        return {
            column: np.nan
            for column in output_columns
        }

    #Copy the aggregated mean and best results into the race features.
    features = {
        column: latest_test[column]
        for column in measurement_columns
    }

    features[days_since_feature] = days_since_test

    return features
```

```python
##Add most recent skills result to each race

def create_skill_features_for_race(race_row):
    """
    Match one race to the athlete's latest relevant start and turn tests.
    """

    race_stroke = race_row["RaceEventSwimmingStyleName"]
    race_distance = race_row["RaceEventDistance"]

    #Butterfly start data used when predicting a medley race.
    start_stroke = (
        "Butterfly"
        if race_stroke == "Medley"
        else race_stroke
    )

    #Find the athlete's latest start test using the appropriate stroke.
    features = find_latest_skill_test(
        test_data=skills_start_tests,
        athlete_name=race_row["AthleteName"],
        race_date=race_row["Date"],
        target_stroke=start_stroke,
        feature_prefix="start",
    )

    #Don't assign turn times for 50m races
    if race_distance > 50:

        #For races longer than 50m, match the latest turn test to the
        #race stroke. Medley races use the combined Medley test label.
        turn_features = find_latest_skill_test(
            test_data=skills_turn_tests,
            athlete_name=race_row["AthleteName"],
            race_date=race_row["Date"],
            target_stroke=race_stroke,
            feature_prefix="turn",
        )

    else:
        #Identify all turn measurement columns so they can be marked
        #unavailable for a 50 m long-course race.
        turn_identifier_columns = {
            "Name",
            "SkillsDate",
            "SkillStroke",
        }

        turn_measurement_columns = [
            column
            for column in skills_turn_tests.columns
            if column not in turn_identifier_columns
        ]

        turn_features = {
            column: np.nan
            for column in turn_measurement_columns
        }

        turn_features["turn_days_since_test"] = np.nan

    #Add the turn features to the start-feature dictionary.
    features.update(turn_features)

    return pd.Series(features)


#Apply the start/turn matching process separately to every race.
skill_features = df_model.apply(
    create_skill_features_for_race,
    axis=1,
)


#Add the resulting skill columns beside all existing model features.
df_model = pd.concat(
    [
        df_model,
        skill_features,
    ],
    axis=1,
)


print(f"Dataset after adding skill features: {df_model.shape}")

#Count races with usable recent start and turn assessments.
print(
    "Races with start-test coverage:",
    df_model["start_10m_out_best"].notna().sum()
)

print(
    "Races with turn-test coverage:",
    df_model["turn_5in_10out_time_best"].notna().sum()
)

#Show how old the matched skill assessments were.
print("\nDays since matched start test:")

print(
    df_model["start_days_since_test"]
    .dropna()
    .describe()
)

print("\nDays since matched turn test:")

print(
    df_model["turn_days_since_test"]
    .dropna()
    .describe()
)
```

```python
##Feature engineering for core 400m test features

#Calculate performance relative to the prescribed target.

#0 means the athlete matched the target.
#+2 means the test was 2% slower than target / -2 means the test was 2% faster than target.

df_400["test_400_pct_from_target"] = np.where(
    df_400["Target Time Seconds"] > 0,
    (
        (
            df_400["Test Time Seconds"]
            / df_400["Target Time Seconds"]
        ) - 1
    ) * 100,
    np.nan,
)


#Calculate the mean stroke count across the two points.

df_400["test_400_stroke_count_mean"] = (
    df_400[
        [
            "Stroke Count Check 1",
            "Stroke Count Check 2",
        ]
    ]
    .mean(
        axis=1,
        skipna=False,     #requires both checks to be present, if either is missing, mean is missing
    )
)


#Calculate how stroke count changed during the test.

#A positive value means stroke count was higher at the second check,
#potentially indicating reduced stroke length or increasing fatigue.
df_400["test_400_stroke_count_change"] = (
    df_400["Stroke Count Check 2"]
    - df_400["Stroke Count Check 1"]
)

#Use the already-numeric RPE version created during preprocessing.
df_400["test_400_rpe"] = df_400["RPE Value"]

#Rename heart rate into a clearer modelling feature without changing the original raw column.
df_400["test_400_heart_rate"] = df_400["Heart Rate"]


TEST_400_FEATURES = [
    "test_400_pct_from_target",
    "test_400_heart_rate",
    "test_400_rpe",
    "test_400_stroke_count_mean",
    "test_400_stroke_count_change",
]


#Inspect distributions and missingness before matching tests to races.
print(
    df_400[TEST_400_FEATURES]
    .describe()
    .T
)

print(
    "\nMissing values:"
)

print(
    df_400[TEST_400_FEATURES]
    .isna()
    .sum()
)
```

```python
##Define pre-race 400m function

def create_400m_test_features(
    test_data,
    athlete_name,
    race_date,
):
    """
    Match one race to the athlete's latest valid weekly 400 m test.
    """

    #Convert the race date into a normalised calendar date.
    race_day = pd.Timestamp(race_date).normalize()

    output_features = [
        *TEST_400_FEATURES,
        "test_400_days_since_test",
    ]

    #Keep tests belonging to this athlete and completed before the race.
    
    #Same-day tests are excluded
    previous_tests = test_data[
        (test_data["About"] == athlete_name)
        & (test_data["Test400Date"] < race_day)
    ]

    #If the athlete has no test before the race (within 21d), all features are missing.
    if previous_tests.empty:
        return {
            feature: np.nan
            for feature in output_features
        }

    #Select the most recent test completed before the race.
    latest_test = (
        previous_tests
        .sort_values("Test400Date")
        .iloc[-1]
    )

    #Calculate how old the test was on the race date.
    days_since_test = (
        race_day - latest_test["Test400Date"]
    ).days

    #Reject tests older than the 21-day limit specified in the settings.
    if days_since_test > TEST_400_MAX_AGE_DAYS:
        return {
            feature: np.nan
            for feature in output_features
        }

    #Copy the selected measures from the latest test into the race-level features.
    features = {
        feature: latest_test[feature]
        for feature in TEST_400_FEATURES
    }

    features["test_400_days_since_test"] = days_since_test

    return features
```

```python
##Apply 400m features to races

def create_400m_features_for_race(race_row):
    """
    Extract one race's athlete and date, then find the athlete's
    most recent valid weekly 400 m test.
    """

    return pd.Series(
        create_400m_test_features(
            #Use the prepared weekly 400 m test dataset.
            test_data=df_400,

            #Match 400 m "About" to Nemo "AthleteName".
            athlete_name=race_row["AthleteName"],

            #Use the race date as the end point for matching.
            race_date=race_row["Date"],
        )
    )


#Apply the matching function separately to every race.
test_400_features = df_model.apply(
    create_400m_features_for_race,
    axis=1,
)


#Add the latest 400 m test features to the race-level dataframe.
df_model = pd.concat(
    [
        df_model,
        test_400_features,
    ],
    axis=1,
)


print(f"Dataset after adding 400 m test features: {df_model.shape}")

#Count races with a valid recent test.
print(
    "Races with recent 400 m test coverage:",
    df_model["test_400_pct_from_target"].notna().sum()
)

#Show the age of the tests linked to races.
print(
    "\nDays between 400 m test and race:"
)

print(
    df_model["test_400_days_since_test"]
    .dropna()
    .describe()
)

#Show missingness among races that were matched to a recent test.
matched_400m_races = df_model[
    df_model["test_400_days_since_test"].notna()
]

print(
    "\nMissing feature values among matched races:"
)

print(
    matched_400m_races[TEST_400_FEATURES]
    .isna()
    .sum()
)
```

```python
##Load/Inspect Teambuildr file

#Load the export.
df_lift = pd.read_csv(
    LIFT_FILE,
    low_memory=False
)

df_lift = df_lift.copy()

#Convert training date into a normalised calendar date.
df_lift["LiftDate"] = (
    pd.to_datetime(
        df_lift["Date"],
        format="mixed",
        dayfirst=True,
        errors="coerce"
    )
    .dt.normalize()
)

#Remove accidental spaces 
df_lift["About"] = (
    df_lift["About"]
    .astype("string")
    .str.strip()
)

print(f"Set-level gym dataset: {df_lift.shape}")
print(f"Missing dates: {df_lift['LiftDate'].isna().sum()}")
print(f"Athletes: {df_lift['About'].nunique()}")
print(f"Gym sessions: {df_lift['event-uuid'].nunique()}")

print(
    "Date range:",
    df_lift["LiftDate"].min().date(),
    "to",
    df_lift["LiftDate"].max().date(),
)
```

```python
##Create gym session dataframe (one row per session)

#Collapse individual set rows into one row per athlete gym session.

#event-uuid identifies the session, while each original row represents one completed set within that session.
lift_sessions = (
    df_lift
    .groupby(
        [
            "event-uuid",
            "About",
            "LiftDate",
        ],
        as_index=False
    )
    .agg(
        #Number of recorded sets in the session
        lift_session_sets=("Set", "size"),

        #Number of distinct exercises completed
        lift_session_exercises=("Completion ID", "nunique"),

        #Sum volume load across sets where volume was measurable
        calculated_volume_load=(
            "Set Volume Load",
            lambda values: values.sum(min_count=1)
        ),

        #TeamBuildr's reported session-level volume total
        reported_volume_load=("Daily Volume Load", "max"),
    )
)


#Prefer TeamBuildr's session total. If it is unavailable, use the
#calculated sum of set-level volume loads.
lift_sessions["lift_session_volume_load"] = (
    lift_sessions["reported_volume_load"]
    .fillna(lift_sessions["calculated_volume_load"])
)


print(f"Session-level gym dataset: {lift_sessions.shape}")

print(
    "Sessions missing volume load:",
    lift_sessions["lift_session_volume_load"]
    .isna()
    .sum()
)
```

```python
##Define rolling features function

def create_lift_features(
    session_data,
    athlete_name,
    race_date,
):
    """
    Calculate recent gym training exposure before one race.
    """

    #Convert race date into a normalised calendar date.
    race_day = pd.Timestamp(race_date).normalize()

    #Keep only gym sessions belonging to this athlete.
    athlete_sessions = session_data[
        session_data["About"] == athlete_name
    ].copy()

    features = {}

    for days in LIFT_WINDOWS:

        #Define beginning of the pre-race window.
        #Gym sessions on the race date itself are excluded.
        window_start = (
            race_day - pd.Timedelta(days=days)
        )

        session_feature = f"lift_sessions_{days}d"
        set_feature = f"lift_sets_{days}d"
        volume_feature = f"lift_volume_load_{days}d"

        #If the athlete is absent or their recorded gym history does not
        #reach the start of the window, it is incomplete.
        if (
            athlete_sessions.empty
            or athlete_sessions["LiftDate"].min() > window_start
        ):
            features[session_feature] = np.nan
            features[set_feature] = np.nan
            features[volume_feature] = np.nan
            continue

        #Select gym sessions within the window and before the race date.
        window_sessions = athlete_sessions[
            (athlete_sessions["LiftDate"] >= window_start)
            & (athlete_sessions["LiftDate"] < race_day)
        ]

        #Count unique sessions so two sessions on the same day remain
        #two separate training exposures.
        number_of_sessions = (
            window_sessions["event-uuid"].nunique()
        )

        features[session_feature] = number_of_sessions

        if number_of_sessions == 0:
            #A fully observed window with no gym sessions is a genuine
            #zero rather than missing data.
            features[set_feature] = 0
            features[volume_feature] = 0

        else:
            #Sum the number of sets completed across all sessions.
            features[set_feature] = (
                window_sessions["lift_session_sets"].sum()
            )

            #Sum quantifiable volume load. If every included session
            #has missing volume, retain NaN rather than returning zero.
            features[volume_feature] = (
                window_sessions["lift_session_volume_load"]
                .sum(min_count=1)
            )

    #Compare the latest seven-day volume with the average weekly volume represented by the 28-day period.
    volume_7d = features.get("lift_volume_load_7d")
    volume_28d = features.get("lift_volume_load_28d")

    if (
        pd.notna(volume_7d)
        and pd.notna(volume_28d)
        and volume_28d > 0
    ):
        features["lift_recent_volume_ratio"] = (
            volume_7d / (volume_28d / 4)
        )
    else:
        features["lift_recent_volume_ratio"] = np.nan

    return features
```

```python
##Apply features to race dataframe

def create_lift_features_for_race(race_row):
    """
    Extract one race's athlete and date, then calculate the relevant
    pre-race gym-load features.
    """

    return pd.Series(
        create_lift_features(
            session_data=lift_sessions,
            athlete_name=race_row["AthleteName"],
            race_date=race_row["Date"],
        )
    )


#Calculate gym-load features separately for every race.
lift_features = df_model.apply(
    create_lift_features_for_race,
    axis=1,
)


#Remove existing gym-feature columns before adding the new versions.

#This makes the cell safe to rerun without creating duplicate columns.
df_model = df_model.drop(
    columns=lift_features.columns,
    errors="ignore",
)


#Add the newly calculated features to the race-level dataframe.
df_model = pd.concat(
    [
        df_model,
        lift_features,
    ],
    axis=1,
)


print(f"Dataset after adding gym-load features: {df_model.shape}")

print(
    "Races with available gym history:",
    df_model["lift_sessions_28d"].notna().sum()
)

print(
    "Races with a gym session in the previous 7 days:",
    (df_model["lift_sessions_7d"] > 0).sum()
)

print(
    "Races with a gym session in the previous 28 days:",
    (df_model["lift_sessions_28d"] > 0).sum()
)

print(
    "Races with a valid recent-volume ratio:",
    df_model["lift_recent_volume_ratio"].notna().sum()
)
```

```python
##Create grouped phases

#Combine sparse race-phase categories into a stable binary context.
    #A-, B- and C-finals are all treated as finals.
    #Heats, semi-finals and trial swims are treated as non-finals.
FINAL_PHASES = [
    "Final",
    "B-Final",
    "C-Final",
]

df_model["RacePhaseGroup"] = np.where(
    df_model["PhaseName"].isin(FINAL_PHASES),
    "Final",
    "Non-final",
)

print(
    df_model["RacePhaseGroup"]
    .value_counts()
)
```

```python
##Common Analysis Cohort

#Confirm every requested predictor and coverage variable exists.
all_required_columns = set(
    [
        OUTCOME_VARIABLE,
        "Date",
        "AthleteName",
        "ModelRaceId",
        *REQUIRED_COVERAGE_VARIABLES,
        *CONTEXT_VARIABLES,
        *INPUT_VARIABLES,])

missing_columns = sorted(
    all_required_columns - set(df_model.columns)
)

if missing_columns:
    raise KeyError(
        "These requested modelling columns do not exist:\n"
        + "\n".join(missing_columns))

#Keep races for which all selected monitoring datasets were
#operating and had sufficient historical coverage.

#Occasional missing measurements inside that coverage are kept for later imputation.
coverage_mask = (
    df_model[REQUIRED_COVERAGE_VARIABLES]
    .notna()
    .all(axis=1))

#Create one common cohort for context-only, monitoring-only and combined
#models. Using the same races, so we can compare performance directly.
analysis_data = df_model.loc[
    coverage_mask,
    [
        "ModelRaceId",
        "Date",
        "AthleteName",
        OUTCOME_VARIABLE,
        *CONTEXT_VARIABLES,
        *INPUT_VARIABLES,
    ],
].copy()

#Remove any accidentally repeated columns
analysis_data = analysis_data.loc[
    :,
    ~analysis_data.columns.duplicated()
]


#Removes any race for which the outcome variable is unexpectedly unavailable.
analysis_data = analysis_data[
    analysis_data[OUTCOME_VARIABLE].notna()
].copy()

#Sort chronologically before constructing validation folds.
analysis_data = (
    analysis_data
    .sort_values(
        [
            "Date",
            "ModelRaceId",])
    .reset_index(drop=True)
)


print(f"Races in common modelling cohort: {len(analysis_data)}")
print(f"Athletes in cohort: {analysis_data['AthleteName'].nunique()}")

print(
    "Cohort date range:",
    analysis_data["Date"].min().date(),
    "to",
    analysis_data["Date"].max().date(),
)

#Examine remaining missingness among the selected monitoring variables.
monitoring_missingness = (
    analysis_data[INPUT_VARIABLES]
    .isna()
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print("\nMonitoring-variable missingness within coverage:")

print(
    monitoring_missingness[
        monitoring_missingness > 0
    ].round(1)
)
```

```python
#--------------------------------
#CROSS-VALIDATION (chronological)
#--------------------------------

def create_chronological_folds(
    data,
    n_splits,
    initial_train_fraction,
):
    """
    Create expanding-window cross-validation folds.

    Every validation period occurs after its corresponding training
    period, and races on the same date remain in the same fold.
    """

    #Identify all distinct race dates in chronological order.
    unique_dates = np.array(
        sorted(data["Date"].dropna().unique())
    )

    number_of_dates = len(unique_dates)

    #Use the earliest 40% of dates for the first training period.
    initial_train_dates = max(
        1,
        int(np.ceil(
            number_of_dates * initial_train_fraction
        ))
    )

    #The remaining dates are divided into five consecutive validation periods/folds.
    validation_dates = unique_dates[
        initial_train_dates:
    ]

    validation_date_blocks = np.array_split(       #Keeps dates in chronological order, divides them evenly.
        validation_dates,
        n_splits,
    )

    folds = []       #Stores the row numbers used for training/validation
    fold_summary = []            #Stores info for inspection and reporting

    for fold_number, validation_block in enumerate(             #provides "fold number" and "validation block"
        validation_date_blocks,
        start=1,              #Starts folding at 1 rather than 0
    ):
        if len(validation_block) == 0:
            raise ValueError(
                "There are not enough distinct race dates "
                "to create the requested number of folds."
            )

        # Training contains every race before this validation block.
        validation_start = validation_block[0]

        train_indices = data.index[
            data["Date"] < validation_start
        ].to_numpy()

        validation_indices = data.index[
            data["Date"].isin(validation_block)
        ].to_numpy()

        folds.append(
            (
                train_indices,
                validation_indices,
            )
        )

        # Store fold details for checking and reporting.
        fold_summary.append({
            "fold": fold_number,
            "training_races": len(train_indices),
            "validation_races": len(validation_indices),
            "training_athletes":
                data.loc[train_indices, "AthleteName"].nunique(),
            "validation_athletes":
                data.loc[validation_indices, "AthleteName"].nunique(),
            "training_start":
                data.loc[train_indices, "Date"].min(),
            "training_end":
                data.loc[train_indices, "Date"].max(),
            "validation_start":
                data.loc[validation_indices, "Date"].min(),
            "validation_end":
                data.loc[validation_indices, "Date"].max(),
        })

    return folds, pd.DataFrame(fold_summary)


chronological_folds, fold_summary = (
    create_chronological_folds(
        data=analysis_data,
        n_splits=CV_FOLDS,
        initial_train_fraction=INITIAL_TRAIN_FRACTION,
    )
)


print(f"Distinct race dates: {analysis_data['Date'].nunique()}")

fold_summary
```

```python
##Import modelling tools, define PREPROCESSING
#Each feature specification will build it's own preprocessing pipeline. Ensures that imputation
#and scaling are learned separately inside each chronological training fold

#-------------------
#MODEL PREPROCESSING
#-------------------

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from sklearn.linear_model import LinearRegression
from sklearn.neural_network import MLPRegressor
from sklearn.dummy import DummyRegressor

from sklearn.compose import TransformedTargetRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
)


def create_preprocessor(
    data,
    feature_variables,
    scale_numeric=False,
):
    """
    Create preprocessing steps for one model specification.

    Numerical and categorical columns are identified from the selected
    feature list rather than from the entire dataframe.
    """

    #Select only the columns entering this model.
    selected_data = data[feature_variables]

    #Identify numeric predictors such as training volume, CMJ measures
    #and race distance.
    numeric_columns = (
        selected_data
        .select_dtypes(include="number")
        .columns
        .tolist()
    )

    #Any selected predictor not recognised as numeric is treated as categorical (e.g. stroke / race phase)
    categorical_columns = [
        column
        for column in feature_variables
        if column not in numeric_columns
    ]


    #Median imputation is fitted using only the training portion of
    #each fold, to prevent validation information influencing it.
    
    numeric_steps = [
        (
            "imputer",
            SimpleImputer(
                strategy="median",
                add_indicator=True,       #creates an extra binary feature to indicate if each value was missing/imputed or not
                keep_empty_features=True,
            ),
        )]


    #MLP requires scaled numerical inputs (variables like metres/force/percentages use different numerical ranges).
    
    #OLS does not require scaling for prediction, so scaling is added only when scale_numeric=True.
    if scale_numeric:
        numeric_steps.append(
            (
                "scaler",
                StandardScaler(),
            )
        )


    numeric_transformer = Pipeline(
        steps=numeric_steps
    )


    #Missing categorical values are replaced by the most frequent training-fold category.
    
    #One-hot encoding converts labels (e.g Freestyle/Butterfly) into numeric indicator columns.
    
    categorical_transformer = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            ),
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",      #prevents error if validation contains category absent from training
                drop=None,
                sparse_output=False,
            ),
        ),
    ])


    #ColumnTransformer applies the appropriate preprocessing separately to numeric and categorical variables.
    preprocessor = ColumnTransformer([
        (
            "numeric",
            numeric_transformer,
            numeric_columns,
        ),
        (
            "categorical",
            categorical_transformer,
            categorical_columns,
        ),
    ])


    return preprocessor
```

```python
##Define the three regression models

#-----------------
#MODEL DEFINITIONS
#-----------------

def create_model_pipelines(
    data,
    feature_variables,
):
    """
    Create baseline, OLS and MLP pipelines for one feature specification.
    """

    #The baseline predicts the mean outcome from each training fold.
        #It provides a minimum benchmark: the more complex models should outperform this
    baseline_pipeline = Pipeline([
        (
            "preprocessor",
            create_preprocessor(
                data=data,
                feature_variables=feature_variables,
                scale_numeric=False,
            ),
        ),
        (
            "model",
            DummyRegressor(
                strategy="mean"
            ),
        ),
    ])


    #LinearRegression estimates an ordinary least-squares relationship 
    #between the selected predictors and percentage from season's best.
    ols_pipeline = Pipeline([
        (
            "preprocessor",
            create_preprocessor(
                data=data,
                feature_variables=feature_variables,
                scale_numeric=False,
            ),
        ),
        (
            "model",
            LinearRegression(),
        ),
    ])


    #MLPRegressor models potentially nonlinear relationships and interactions between the selected predictors.
    #Corrections:
        #Use a smaller neural network because the chronological training folds contain only 263–446 races.
        #The target (SB%) is also standardized within each training fold and converted back to the original 
        #percentage scale when predictions are generated.
    mlp_pipeline = Pipeline([
        (
            "preprocessor",
            create_preprocessor(
                data=data,
                feature_variables=feature_variables,
                scale_numeric=True,
            ),
        ),
        (
            "model",
            TransformedTargetRegressor(
                regressor=MLPRegressor(
                    #smaller network reduces parameters relative to sample size
                    hidden_layer_sizes=(16,8),

                    #L2 regularisation reduces large weights
                    alpha = 0.01,

                    max_iter = 2000,

                    early_stopping = True,    #stop when performance no longer imrpoves (after 30)
                    validation_fraction = 0.15,
                    n_iter_no_change = 30,

                    random_state = RANDOM_STATE,
                ),

                #Standardise outcome using training-fold values only. Predictions then returned to original
                transformer=StandardScaler(),
            ),
        ),
    ])


    return {
        "Baseline": baseline_pipeline,
        "OLS": ols_pipeline,
        "MLP": mlp_pipeline,
    }
```

```python
##Run all the models across the five folds (Fits 45 models)
##3 feature specifications × 3 models × 5 folds

#----------------------------------
#RUN CHRONOLOGICAL CROSS-VALIDATION
#----------------------------------

from sklearn.base import clone      #explained later


#Store one row of evaluation results for every fitted model and fold.
fold_results = []

#Store individual validation predictions
validation_predictions = []

#Work through the context-only, monitoring-only and combined feature specifications separately.
for feature_set_name, feature_variables in MODEL_FEATURE_SETS.items():

    print(f"\nRunning feature set: {feature_set_name}")

    #Create unfitted model templates using the preprocessing appropriate for this feature specification.
    model_templates = create_model_pipelines(
        data=analysis_data,
        feature_variables=feature_variables,
    )

    #Separate the selected predictors from the outcome.
    X = analysis_data[feature_variables]
    y = analysis_data[OUTCOME_VARIABLE]


    #Run the baseline, OLS and MLP models.
    for model_name, model_template in model_templates.items():

        print(f"  Model: {model_name}")

        #Run the model once for each chronological fold.
        for fold_number, (
            train_indices,
            validation_indices,
        ) in enumerate(
            chronological_folds,
            start=1,
        ):

            #Select all earlier races used to train this fold.
            X_train = X.loc[train_indices]
            y_train = y.loc[train_indices]

            #Select the later races used to validate this fold.
            X_validation = X.loc[validation_indices]
            y_validation = y.loc[validation_indices]


            #clone() creates a new unfitted copy of the pipeline.
                #prevents information learned in one fold from carrying into the next fold.
            fitted_model = clone(model_template)


            #Fit preprocessing and the regression model using only the training races from this fold.
            fitted_model.fit(
                X_train,
                y_train,)


            #Generate predictions for the later validation races.
            predictions = fitted_model.predict(
                X_validation)


            #Calculate fold-level evaluation metrics.
            fold_mae = mean_absolute_error(
                y_validation,
                predictions,
            )

            fold_rmse = (
                mean_squared_error(
                    y_validation,
                    predictions,
                ) ** 0.5
            )

            fold_r2 = r2_score(
                y_validation,
                predictions,
            )


            fold_results.append({
                "feature_set": feature_set_name,
                "model": model_name,
                "fold": fold_number,
                "training_races": len(train_indices),
                "validation_races": len(validation_indices),
                "MAE": fold_mae,
                "RMSE": fold_rmse,
                "R2": fold_r2,
            })


            #Keep every individual prediction so overall performance can
            #later be calculated across all validation races together.
            fold_prediction_data = analysis_data.loc[
                validation_indices,
                [
                    "ModelRaceId",
                    "Date",
                    "AthleteName",
                ],
            ].copy()

            fold_prediction_data["actual"] = (
                y_validation.to_numpy()
            )

            fold_prediction_data["predicted"] = predictions
            fold_prediction_data["feature_set"] = feature_set_name
            fold_prediction_data["model"] = model_name
            fold_prediction_data["fold"] = fold_number

            validation_predictions.append(
                fold_prediction_data
            )


#Convert the stored results into dataframes.
fold_results = pd.DataFrame(fold_results)

validation_predictions = pd.concat(
    validation_predictions,
    ignore_index=True,
)


print(
    "\nCompleted model fits:",
    len(fold_results)
)

print(
    "Stored validation predictions:",
    len(validation_predictions)
)
```

```python
#--------------------------------------
#Summarise Model Validation Performance
#--------------------------------------

#Calculate average and variability of performance across the five chronological validation folds.
#Shows whether a model performs consistently over time.
fold_performance_summary = (
    fold_results
    .groupby(
        [
            "feature_set",
            "model",
        ]
    )
    .agg(
        folds=("fold", "count"),

        mean_MAE=("MAE", "mean"),
        sd_MAE=("MAE", "std"),

        mean_RMSE=("RMSE", "mean"),
        sd_RMSE=("RMSE", "std"),

        mean_R2=("R2", "mean"),
        sd_R2=("R2", "std"),
    )
    .reset_index()
)


#Calculate performance across all chronological validation predictions pooled together.

#Pooled metrics are useful because validation folds contain different
#numbers of races. Each validation race contributes equally here,
#rather than each fold receiving equal weight.
pooled_results = []

for (
    feature_set_name,
    model_name,
), prediction_group in validation_predictions.groupby(
    [
        "feature_set",
        "model",
    ]
):

    actual_values = prediction_group["actual"]
    predicted_values = prediction_group["predicted"]

    pooled_results.append({
        "feature_set": feature_set_name,
        "model": model_name,
        "validation_races": len(prediction_group),

        "pooled_MAE": mean_absolute_error(
            actual_values,
            predicted_values,
        ),

        "pooled_RMSE": (
            mean_squared_error(
                actual_values,
                predicted_values,
            ) ** 0.5
        ),

        "pooled_R2": r2_score(
            actual_values,
            predicted_values,
        ),

        #Positive bias:  model generally predicts a higher percentage (slower performance) than observed.
        "prediction_bias": (
            predicted_values - actual_values
        ).mean(),
    })


pooled_performance_summary = (
    pd.DataFrame(pooled_results)
    .sort_values(
        [
            "feature_set",
            "pooled_MAE",
        ]
    )
    .reset_index(drop=True)
)


print("Average performance across folds:")

display(
    fold_performance_summary.round(3)
)


print("\nPerformance pooled across all validation races:")

display(
    pooled_performance_summary.round(3)
)
```
